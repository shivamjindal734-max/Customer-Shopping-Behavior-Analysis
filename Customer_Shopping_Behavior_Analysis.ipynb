```python
import pandas as pd

df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating'] = df['Review Rating'].fillna(df.groupby('Category')['Review Rating'].transform('median'))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns
```

```python
df['age_group'] = pd.qcut(df['age'], 4, labels=['Young Adult', 'Adult', 'Middle-aged', 'Senior'])
```

```python
df[['age','age_group']].head()
```

```python
print(df.columns)
```

```python
df['purchase_frequency_days'] = df['frequency of purchases'].replace(
    ['Weekly', 'Bi-Weekly', 'Fortnightly', 'Monthly', 'Quarterly', 'Every 3 Months', 'Annually'], 
    [7, 14, 14, 30, 90, 90, 365]
)
```

```python
df[['purchase_frequency_days','frequency of purchases']].head(30)
```

```python
df[['discount applied','promo code used']].head()
```

```python
df.columns
```

```python
df.to_csv("cleaned_data.csv", index=False)
```
